```python
import pandas as pd
```

```python
data = {
    "Name": ["Ravi", "Sita", "Raju", "Priya"],
    "Age": [20, 21, 19, 20],
    "Marks": [85, 90, 75, 95]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.to_excel("students.xlsx", index=False)
```

```python
df = pd.read_excel("students.xlsx")

print(df)
```

```python
print(df.head())
print(df.shape)
print(df.info())
print(df.describe())
```

```python
df = df.dropna()
```

```python
df["Result"] = df["Marks"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)

print(df)
```

```python
df.to_excel("processed_students.xlsx", index=False)
```
